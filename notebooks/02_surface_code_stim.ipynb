```python
circuit = stim.Circuit.generated(
    "surface_code:rotated_memory_z",
    distance=3,
    rounds=3,
)   # every noise parameter defaults to 0, so this is noiseless

print(circuit.num_qubits, circuit.num_measurements,
      circuit.num_detectors, circuit.num_observables)
```

```python
coords = circuit.get_final_qubit_coordinates()
print(len(coords))           # number of qubits actually used
for q, (x, y) in sorted(coords.items()):
    print(q, (x, y))
```
